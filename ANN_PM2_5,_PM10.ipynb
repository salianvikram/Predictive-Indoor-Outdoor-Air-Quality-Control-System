```python
# ============================================================
# AQI PREDICTION USING ARTIFICIAL NEURAL NETWORK (ANN)
#
# DATASET:
# station_hour_pm_aqi.csv
#
# AVAILABLE COLUMNS:
# Datetime
# PM2.5
# PM10
# AQI
#
# MODEL:
# Artificial Neural Network
#
# TARGET:
# NEXT-HOUR AQI
#
# OPTIMIZED FOR:
# Google Colab
# Large Dataset
# Lower RAM Usage
# Faster Training
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib
import gc

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Dense,
    Dropout,
    Input
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)


print("Libraries imported successfully.")


# ============================================================
# 2. MEMORY CLEANUP
# ============================================================

tf.keras.backend.clear_session()

gc.collect()

print("Memory cleanup completed.")


# ============================================================
# 3. CHECK GPU
# ============================================================

gpus = tf.config.list_physical_devices("GPU")

if len(gpus) > 0:

    print(
        "\nGPU detected:"
    )

    print(gpus)

else:

    print(
        "\nGPU not detected."
    )

    print(
        "ANN can still run on CPU."
    )


# ============================================================
# 4. LOAD DATASET
# ============================================================

FILE_NAME = "station_hour_pm_aqi.csv"


df = pd.read_csv(

    FILE_NAME,

    dtype={

        "PM2.5": np.float32,

        "PM10": np.float32,

        "AQI": np.float32

    }

)


print(
    "\nDataset loaded successfully."
)


print(
    "Dataset shape:",
    df.shape
)


# ============================================================
# 5. DISPLAY DATASET INFORMATION
# ============================================================

print(
    "\nDataset columns:"
)


print(
    df.columns.tolist()
)


print(
    "\nFirst 5 rows:"
)


print(
    df.head()
)


# ============================================================
# 6. CLEAN COLUMN NAMES
# ============================================================

df.columns = (

    df.columns

    .str.strip()

)


# ============================================================
# 7. CHECK REQUIRED COLUMNS
# ============================================================

required_columns = [

    "Datetime",

    "PM2.5",

    "PM10",

    "AQI"

]


missing_columns = [

    col

    for col in required_columns

    if col not in df.columns

]


if len(missing_columns) > 0:

    raise ValueError(

        "Missing required columns: "

        + str(missing_columns)

    )


print(
    "\nAll required columns are available."
)


# ============================================================
# 8. CONVERT DATETIME
# ============================================================

df["Datetime"] = pd.to_datetime(

    df["Datetime"],

    format="%d-%m-%Y %H:%M",

    errors="coerce"

)


# Remove invalid dates

df = df.dropna(

    subset=["Datetime"]

)


print(
    "\nDatetime conversion completed."
)


# ============================================================
# 9. CONVERT NUMERIC COLUMNS
# ============================================================

numeric_columns = [

    "PM2.5",

    "PM10",

    "AQI"

]


for col in numeric_columns:

    df[col] = pd.to_numeric(

        df[col],

        errors="coerce"

    ).astype(

        np.float32

    )


print(
    "\nNumeric conversion completed."
)


# ============================================================
# 10. SORT DATA CHRONOLOGICALLY
# ============================================================

df = (

    df

    .sort_values(

        by="Datetime"

    )

    .reset_index(

        drop=True

    )

)


print(
    "\nDataset sorted chronologically."
)


# ============================================================
# 11. HANDLE MISSING VALUES
# ============================================================

for col in numeric_columns:

    df[col] = (

        df[col]

        .ffill()

        .bfill()

    )


print(
    "\nMissing values handled."
)


# ============================================================
# 12. CREATE TIME FEATURES
# ============================================================

df["Hour"] = (

    df["Datetime"]

    .dt.hour

)


df["Day"] = (

    df["Datetime"]

    .dt.day

)


df["Month"] = (

    df["Datetime"]

    .dt.month

)


df["DayOfWeek"] = (

    df["Datetime"]

    .dt.dayofweek

)


df["DayOfYear"] = (

    df["Datetime"]

    .dt.dayofyear

)


df["Year"] = (

    df["Datetime"]

    .dt.year

)


df["IsWeekend"] = (

    df["DayOfWeek"] >= 5

).astype(

    np.int8

)


print(
    "\nTime features created."
)


# ============================================================
# 13. CREATE CYCLIC TIME FEATURES
# ============================================================

df["Hour_Sin"] = (

    np.sin(

        2 * np.pi *

        df["Hour"] / 24

    )

).astype(

    np.float32

)


df["Hour_Cos"] = (

    np.cos(

        2 * np.pi *

        df["Hour"] / 24

    )

).astype(

    np.float32

)


df["Month_Sin"] = (

    np.sin(

        2 * np.pi *

        df["Month"] / 12

    )

).astype(

    np.float32

)


df["Month_Cos"] = (

    np.cos(

        2 * np.pi *

        df["Month"] / 12

    )

).astype(

    np.float32

)


df["DayOfWeek_Sin"] = (

    np.sin(

        2 * np.pi *

        df["DayOfWeek"] / 7

    )

).astype(

    np.float32

)


df["DayOfWeek_Cos"] = (

    np.cos(

        2 * np.pi *

        df["DayOfWeek"] / 7

    )

).astype(

    np.float32

)


print(
    "\nCyclic time features created."
)


# ============================================================
# 14. CREATE PM2.5 LAG FEATURES
# ============================================================

lag_hours = [

    1,

    3,

    6,

    12,

    24

]


for lag in lag_hours:

    df[

        f"PM2.5_lag_{lag}"

    ] = (

        df["PM2.5"]

        .shift(lag)

        .astype(

            np.float32

        )

    )


print(
    "\nPM2.5 lag features created."
)


# ============================================================
# 15. CREATE PM10 LAG FEATURES
# ============================================================

for lag in lag_hours:

    df[

        f"PM10_lag_{lag}"

    ] = (

        df["PM10"]

        .shift(lag)

        .astype(

            np.float32

        )

    )


print(
    "\nPM10 lag features created."
)


# ============================================================
# 16. CREATE AQI LAG FEATURES
# ============================================================

for lag in lag_hours:

    df[

        f"AQI_lag_{lag}"

    ] = (

        df["AQI"]

        .shift(lag)

        .astype(

            np.float32

        )

    )


print(
    "\nAQI lag features created."
)


# ============================================================
# 17. CREATE PM2.5 ROLLING FEATURES
# ============================================================

df["PM2.5_rolling_3"] = (

    df["PM2.5"]

    .shift(1)

    .rolling(

        window=3,

        min_periods=3

    )

    .mean()

    .astype(

        np.float32

    )

)


df["PM2.5_rolling_6"] = (

    df["PM2.5"]

    .shift(1)

    .rolling(

        window=6,

        min_periods=6

    )

    .mean()

    .astype(

        np.float32

    )

)


df["PM2.5_rolling_24"] = (

    df["PM2.5"]

    .shift(1)

    .rolling(

        window=24,

        min_periods=24

    )

    .mean()

    .astype(

        np.float32

    )

)


print(
    "\nPM2.5 rolling features created."
)


# ============================================================
# 18. CREATE PM10 ROLLING FEATURES
# ============================================================

df["PM10_rolling_3"] = (

    df["PM10"]

    .shift(1)

    .rolling(

        window=3,

        min_periods=3

    )

    .mean()

    .astype(

        np.float32

    )

)


df["PM10_rolling_6"] = (

    df["PM10"]

    .shift(1)

    .rolling(

        window=6,

        min_periods=6

    )

    .mean()

    .astype(

        np.float32

    )

)


df["PM10_rolling_24"] = (

    df["PM10"]

    .shift(1)

    .rolling(

        window=24,

        min_periods=24

    )

    .mean()

    .astype(

        np.float32

    )

)


print(
    "\nPM10 rolling features created."
)


# ============================================================
# 19. CREATE AQI ROLLING FEATURES
# ============================================================

df["AQI_rolling_3"] = (

    df["AQI"]

    .shift(1)

    .rolling(

        window=3,

        min_periods=3

    )

    .mean()

    .astype(

        np.float32

    )

)


df["AQI_rolling_6"] = (

    df["AQI"]

    .shift(1)

    .rolling(

        window=6,

        min_periods=6

    )

    .mean()

    .astype(

        np.float32

    )

)


df["AQI_rolling_24"] = (

    df["AQI"]

    .shift(1)

    .rolling(

        window=24,

        min_periods=24

    )

    .mean()

    .astype(

        np.float32

    )

)


print(
    "\nAQI rolling features created."
)


# ============================================================
# 20. CREATE NEXT-HOUR AQI TARGET
# ============================================================

# Current observation:
#
# t
#
# Target:
#
# t + 1 hour


df["AQI_Next_Hour"] = (

    df["AQI"]

    .shift(-1)

    .astype(

        np.float32

    )

)


print(
    "\nNext-hour AQI target created."
)


# ============================================================
# 21. DEFINE FEATURE LIST
# ============================================================

features = [

    # --------------------------------------------------------
    # Current pollutant values
    # --------------------------------------------------------

    "PM2.5",

    "PM10",


    # --------------------------------------------------------
    # Time features
    # --------------------------------------------------------

    "Hour",

    "Day",

    "Month",

    "DayOfWeek",

    "DayOfYear",

    "Year",

    "IsWeekend",


    # --------------------------------------------------------
    # Cyclic features
    # --------------------------------------------------------

    "Hour_Sin",

    "Hour_Cos",

    "Month_Sin",

    "Month_Cos",

    "DayOfWeek_Sin",

    "DayOfWeek_Cos",


    # --------------------------------------------------------
    # PM2.5 lag features
    # --------------------------------------------------------

    "PM2.5_lag_1",

    "PM2.5_lag_3",

    "PM2.5_lag_6",

    "PM2.5_lag_12",

    "PM2.5_lag_24",


    # --------------------------------------------------------
    # PM10 lag features
    # --------------------------------------------------------

    "PM10_lag_1",

    "PM10_lag_3",

    "PM10_lag_6",

    "PM10_lag_12",

    "PM10_lag_24",


    # --------------------------------------------------------
    # AQI lag features
    # --------------------------------------------------------

    "AQI_lag_1",

    "AQI_lag_3",

    "AQI_lag_6",

    "AQI_lag_12",

    "AQI_lag_24",


    # --------------------------------------------------------
    # PM2.5 rolling features
    # --------------------------------------------------------

    "PM2.5_rolling_3",

    "PM2.5_rolling_6",

    "PM2.5_rolling_24",


    # --------------------------------------------------------
    # PM10 rolling features
    # --------------------------------------------------------

    "PM10_rolling_3",

    "PM10_rolling_6",

    "PM10_rolling_24",


    # --------------------------------------------------------
    # AQI rolling features
    # --------------------------------------------------------

    "AQI_rolling_3",

    "AQI_rolling_6",

    "AQI_rolling_24"

]


print(
    "\nTotal number of features:",
    len(features)
)


# ============================================================
# 22. REPLACE INFINITE VALUES
# ============================================================

df = df.replace(

    [np.inf, -np.inf],

    np.nan

)


# ============================================================
# 23. REMOVE MISSING VALUES
# ============================================================

df_model = (

    df

    .dropna(

        subset=

        features
        +
        ["AQI_Next_Hour"]

    )

    .copy()

)


print(
    "\nRows after removing missing values:",
    len(df_model)
)


# ============================================================
# 24. CONVERT FEATURES TO FLOAT32
# ============================================================

df_model[features] = (

    df_model[features]

    .astype(

        np.float32

    )

)


df_model["AQI_Next_Hour"] = (

    df_model["AQI_Next_Hour"]

    .astype(

        np.float32

    )

)


# ============================================================
# 25. TIME-BASED TRAIN / TEST SPLIT
# ============================================================

# Training:
# 2015 - 2019
#
# Testing:
# 2020


train_df = df_model[

    df_model["Year"] < 2020

].copy()


test_df = df_model[

    df_model["Year"] >= 2020

].copy()


print("\n")

print("=" * 60)

print(
    "TRAIN / TEST SPLIT"
)

print("=" * 60)

print(
    "Training rows:",
    len(train_df)
)

print(
    "Testing rows:",
    len(test_df)
)

print("=" * 60)


# ============================================================
# 26. CHECK TRAIN / TEST DATA
# ============================================================

if len(train_df) == 0:

    raise ValueError(

        "Training dataset is empty."

    )


if len(test_df) == 0:

    raise ValueError(

        "Testing dataset is empty."

    )


# ============================================================
# 27. CREATE TRAINING DATA
# ============================================================

X_train = (

    train_df[features]

    .to_numpy(

        dtype=np.float32

    )

)


y_train = (

    train_df["AQI_Next_Hour"]

    .to_numpy(

        dtype=np.float32

    )

    .reshape(

        -1,

        1

    )

)


# ============================================================
# 28. CREATE TESTING DATA
# ============================================================

X_test = (

    test_df[features]

    .to_numpy(

        dtype=np.float32

    )

)


y_test = (

    test_df["AQI_Next_Hour"]

    .to_numpy(

        dtype=np.float32

    )

    .reshape(

        -1,

        1

    )

)


print(
    "\nX_train:",
    X_train.shape
)

print(
    "X_test:",
    X_test.shape
)


# ============================================================
# 29. FREE DATAFRAME MEMORY
# ============================================================

del df

del df_model

del train_df

del test_df

gc.collect()


# ============================================================
# 30. FEATURE SCALING
# ============================================================

feature_scaler = StandardScaler()

target_scaler = StandardScaler()


X_train_scaled = (

    feature_scaler

    .fit_transform(

        X_train

    )

    .astype(

        np.float32

    )

)


X_test_scaled = (

    feature_scaler

    .transform(

        X_test

    )

    .astype(

        np.float32

    )

)


y_train_scaled = (

    target_scaler

    .fit_transform(

        y_train

    )

    .astype(

        np.float32

    )

)


y_test_scaled = (

    target_scaler

    .transform(

        y_test

    )

    .astype(

        np.float32

    )

)


# ============================================================
# 31. FREE UNSCALED DATA
# ============================================================

del X_train

del X_test

del y_train

del y_test

gc.collect()


# ============================================================
# 32. BUILD ANN MODEL
# ============================================================

model = Sequential([

    Input(
        shape=(len(features),)
    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(
        0.15
    ),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(
        0.15
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        1
    )

])


# ============================================================
# 33. COMPILE ANN
# ============================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=0.001

    ),

    loss="mse",

    metrics=["mae"]

)


print("\n")

print("=" * 60)

print(
    "ANN MODEL"
)

print("=" * 60)

model.summary()


# ============================================================
# 34. EARLY STOPPING
# ============================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=5,

    restore_best_weights=True,

    verbose=1

)


# ============================================================
# 35. LEARNING RATE REDUCTION
# ============================================================

reduce_lr = ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=2,

    min_lr=1e-6,

    verbose=1

)


# ============================================================
# 36. TRAIN ANN
# ============================================================

print("\n")

print("=" * 60)

print(
    "TRAINING ANN MODEL..."
)

print("=" * 60)


history = model.fit(

    X_train_scaled,

    y_train_scaled,

    epochs=30,

    batch_size=512,

    validation_split=0.10,

    callbacks=[

        early_stopping,

        reduce_lr

    ],

    verbose=1

)


print(
    "\nANN training completed."
)


# ============================================================
# 37. MAKE PREDICTIONS
# ============================================================

print(
    "\nGenerating predictions..."
)


y_pred_scaled = model.predict(

    X_test_scaled,

    batch_size=512,

    verbose=1

)


print(
    "Predictions generated."
)


# ============================================================
# 38. CONVERT PREDICTIONS BACK TO AQI SCALE
# ============================================================

y_pred = (

    target_scaler

    .inverse_transform(

        y_pred_scaled

    )

    .flatten()

)


y_actual = (

    target_scaler

    .inverse_transform(

        y_test_scaled

    )

    .flatten()

)


# ============================================================
# 39. CALCULATE MAE
# ============================================================

mae = mean_absolute_error(

    y_actual,

    y_pred

)


# ============================================================
# 40. CALCULATE MSE
# ============================================================

mse = mean_squared_error(

    y_actual,

    y_pred

)


# ============================================================
# 41. CALCULATE RMSE
# ============================================================

rmse = np.sqrt(

    mse

)


# ============================================================
# 42. CALCULATE R²
# ============================================================

r2 = r2_score(

    y_actual,

    y_pred

)


# ============================================================
# 43. DISPLAY RESULTS
# ============================================================

print("\n")

print("=" * 60)

print(
    "ANN MODEL PERFORMANCE"
)

print("=" * 60)

print(
    f"MAE  : {mae:.2f}"
)

print(
    f"MSE  : {mse:.2f}"
)

print(
    f"RMSE : {rmse:.2f}"
)

print(
    f"R²   : {r2:.4f}"
)

print("=" * 60)


# ============================================================
# 44. MODEL INFORMATION
# ============================================================

print("\n")

print("=" * 60)

print(
    "MODEL INFORMATION"
)

print("=" * 60)

print(
    "Algorithm       : Artificial Neural Network"
)

print(
    "Target          : Next-Hour AQI"
)

print(
    "Input pollutants: PM2.5, PM10"
)

print(
    "Architecture    : 64 → 32 → 16 → 1"
)

print(
    "Activation      : ReLU"
)

print(
    "Optimizer       : Adam"
)

print(
    "Batch size      : 512"
)

print(
    "Training rows   :",
    len(X_train_scaled)
)

print(
    "Testing rows    :",
    len(X_test_scaled)
)

print(
    "Total features  :",
    len(features)
)

print("=" * 60)


# ============================================================
# 45. TRAINING LOSS GRAPH
# ============================================================

plt.figure(

    figsize=(12, 6)

)


plt.plot(

    history.history["loss"],

    label="Training Loss"

)


plt.plot(

    history.history["val_loss"],

    label="Validation Loss"

)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "MSE Loss"
)

plt.title(
    "ANN Training and Validation Loss"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 46. ACTUAL VS PREDICTED AQI
# ============================================================

sample_size = min(

    500,

    len(y_actual)

)


plt.figure(

    figsize=(12, 6)

)


plt.plot(

    y_actual[
        :sample_size
    ],

    label="Actual AQI"

)


plt.plot(

    y_pred[
        :sample_size
    ],

    label="Predicted AQI"

)


plt.xlabel(
    "Observation"
)

plt.ylabel(
    "AQI"
)

plt.title(
    "ANN - Actual vs Predicted Next-Hour AQI"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 47. PREDICTION ERROR DISTRIBUTION
# ============================================================

errors = (

    y_actual

    -

    y_pred

)


plt.figure(

    figsize=(10, 6)

)


plt.hist(

    errors,

    bins=50

)


plt.xlabel(
    "Prediction Error"
)

plt.ylabel(
    "Frequency"
)

plt.title(
    "ANN Prediction Error Distribution"
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 48. ACTUAL VS PREDICTED SCATTER PLOT
# ============================================================

plt.figure(

    figsize=(8, 8)

)


plt.scatter(

    y_actual,

    y_pred,

    s=5,

    alpha=0.3

)


# Perfect prediction line

min_value = min(

    y_actual.min(),

    y_pred.min()

)


max_value = max(

    y_actual.max(),

    y_pred.max()

)


plt.plot(

    [min_value, max_value],

    [min_value, max_value],

    linestyle="--"

)


plt.xlabel(
    "Actual AQI"
)

plt.ylabel(
    "Predicted AQI"
)

plt.title(
    "Actual vs Predicted AQI - ANN"
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 49. SHOW SAMPLE PREDICTIONS
# ============================================================

comparison = pd.DataFrame({

    "Actual_AQI":

        y_actual[
            :20
        ],

    "Predicted_AQI":

        y_pred[
            :20
        ]

})


comparison["Absolute_Error"] = (

    abs(

        comparison["Actual_AQI"]

        -

        comparison["Predicted_AQI"]

    )

)


print("\n")

print("=" * 60)

print(
    "SAMPLE PREDICTIONS"
)

print("=" * 60)

print(

    comparison

    .round(2)

)


# ============================================================
# 50. SAVE ANN MODEL
# ============================================================

MODEL_PATH = (

    "AQI_ANN_PM_Model.keras"

)


model.save(

    MODEL_PATH

)


print(
    "\nANN model saved successfully:"
)

print(
    MODEL_PATH
)


# ============================================================
# 51. SAVE FEATURE SCALER
# ============================================================

joblib.dump(

    feature_scaler,

    "AQI_ANN_Feature_Scaler.pkl"

)


print(
    "\nFeature scaler saved successfully."
)


# ============================================================
# 52. SAVE TARGET SCALER
# ============================================================

joblib.dump(

    target_scaler,

    "AQI_ANN_Target_Scaler.pkl"

)


print(
    "Target scaler saved successfully."
)


# ============================================================
# 53. SAVE FEATURE LIST
# ============================================================

joblib.dump(

    features,

    "AQI_ANN_Features.pkl"

)


print(
    "Feature list saved successfully."
)


# ============================================================
# 54. FINAL SUMMARY
# ============================================================

print("\n")

print("=" * 60)

print(
    "FINAL ANN MODEL SUMMARY"
)

print("=" * 60)

print(
    "Dataset          : station_hour_pm_aqi.csv"
)

print(
    "Input pollutants : PM2.5, PM10"
)

print(
    "Target           : Next-Hour AQI"
)

print(
    "Model            : Artificial Neural Network"
)

print(
    "Architecture     : 64 → 32 → 16 → 1"
)

print(
    "Lag periods      : 1, 3, 6, 12, 24"
)

print(
    "Rolling periods  : 3, 6, 24"
)

print(
    "Training period  : Before 2020"
)

print(
    "Testing period   : 2020"
)

print(
    f"MAE              : {mae:.2f}"
)

print(
    f"MSE              : {mse:.2f}"
)

print(
    f"RMSE             : {rmse:.2f}"
)

print(
    f"R²               : {r2:.4f}"
)

print("=" * 60)


# ============================================================
# 55. FINAL MEMORY CLEANUP
# ============================================================

del X_train_scaled
del X_test_scaled
del y_train_scaled
del y_test_scaled

gc.collect()


print(
    "\nANN execution completed successfully."
)
```
